# 05 Volatility or structure?

Forbes and Rigobon (2002) showed that measured correlation rises mechanically when volatility rises, so part of any crisis spike is a statistical effect. This notebook estimates how large that part is. For each year it fits a one-factor market model, keeps that year's betas, resets market and idiosyncratic volatility to their typical 2004-2024 levels, and recomputes the average correlation. The adjustment is exact under a one-factor model, and notebook 04 showed this panel is close to one.

At typical volatility the series is flat, between 0.23 and 0.30 in every year, COVID included. None of the rise from 2005 to 2008 survives the adjustment, and 3% of the move from 2017 to 2022 does. The peaks and the low have the same cause: market variance was 6.9 times its typical level in 2008, 7.7 times in 2020 and 3.1 times in 2011, and a quarter of it in 2017.

Volatilities are annualised (variance per observation times observations in the year). An earlier version used per-observation variances, which mixed the sparse pre-2019 years with the complete later ones and made the adjusted series drift upwards after 2019.
